## Exercício 5 — Intermediário
notas = {
    "Ana": 8,
    "Carlos": 5,
    "Marina": 9
}

Faça:

Mostre apenas os alunos com nota ≥ 7

```python
def show_student(name, score):
  print(f">> {name} -> {score} <-")

scores = {
    "Ann" : 8,
    "Carla" : 10,
    "Marina" : 0
}

for student, score in scores.items():
  if score >= 7:
    show_student(student, score)
```

## Exercício 6 — Agrupamento de Dados


dados = [

    {"nome": "Ana", "curso": "Dados"},
    {"nome": "Carlos", "curso": "Dev"},
    {"nome": "Marina", "curso": "Dados"},
    {"nome": "João", "curso": "Dev"}
]

Faça:

Crie um dicionário que agrupe os nomes por curso:

Resultado esperado:

{

    "Dados": ["Ana", "Marina"],
    "Dev": ["Carlos", "João"]
}

```python
data = [
    {"nome": "Ana", "curso": "Dados"},
    {"nome": "Carlos", "curso": "Dev"},
    {"nome": "Marina", "curso": "Dados"},
    {"nome": "João", "curso": "Dev"}
]

dicts = {}
mini_list_1 = []
mini_list_2 = []

for d in data:
  if d["curso"] == "Dados":
    mini_list_1.append(d["nome"])
    dicts[d["curso"]] = mini_list_1
  elif d["curso"] == "Dev":
    mini_list_2.append(d["nome"])
    dicts[d["curso"]] = mini_list_2

print(dicts)

```

## Exercício 7 — Maior valor no dicionário
vendas = {

    "jan": 1500,
    "fev": 2300,
    "mar": 1800,
    "abr": 3000
}

Faça:

Descubra qual mês teve a maior venda
Retorne o nome do mês

Resultado esperado:

    "abr"

```python
sales = {
    "January" : 1500,
    "Febuary" : 2300,
    "March" : 1800,
    "April" : 300
}

bigger_sale_price = 0.0
month_bigger_sale_price = "January"
bigger_sale_price = sales[month_bigger_sale_price]

for month, sale_price in sales.items():
  if sale_price > bigger_sale_price:
    month_sale_price = month
    bigger_sale_price = sale_price


print(f"O mês com maior venda foi: {month_sale_price}: ${bigger_sale_price}")

```
